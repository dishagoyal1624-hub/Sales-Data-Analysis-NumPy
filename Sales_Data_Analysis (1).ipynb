# Sales Data Analysis

## Project Overview

This project focuses on analyzing sales data.

The dataset contains sales information for different products
across different months. The analysis helps in understanding
sales trends, product performance, and overall business performance.

## Objectives

- Analyze sales data
- Calculate total and average sales
- Find highest and lowest sales
- Perform monthly sales analysis
- Perform product-wise sales analysis
- Identify the top-performing product
- Analyze overall sales performance

```python
months = np.array([
    "January", "February", "March", "April", "May", "June",
"July", "August", "September", "October", "November", "December"
])


products = np.array([
    "Laptop", "Smartphone", "Headphones", "Keyboard", "Mouse"
])

sales = np.array([
    [45000,38000,12000,8000,5000], 
    [43500,40500,11800,8700,5300], 
    [48200,39500,13200,8400,5800], 
    [46100,42500,12600,9100,5600], 
    [51500,44800,13900,9300,6200], 
    [53200,47200,15100,9800,6800], 
    [49800,45900,14500,9200,6400],
    [55600,48500,15800,10100,7100], 
    [54100,51200,16200,10500,7300], 
    [62800,58400,18100,11900,8200], 
    [71400,65200,20900,13400,9100],
    [79200,73800,23600,15100,10200]
])
```

## Dataset Description
The dataset contains monthly sales data for five products from January to December.

```python
## Q-1: Calculate Total Sales
```

```python
total_sales = np.sum(sales)
total_sales
```

```python
## Q-2: Calculate Average Sales
```

```python
average_sales = np.average(sales)
average_sales
```

```python
## Q-3: Find the Highest Individual Sale
```

```python
highest_sales = np.max(sales)
highest_index = np.argmax(sales)
row, column = np.unravel_index(np.argmax(sales), sales.shape)
months[row], products[column], sales[row, column]
```

```python
## Q-4: Find the Lowest Individual Sale
```

```python
lowest_sales = np.min(sales)
lowest_index = np.argmin(sales)
row, column = np.unravel_index(np.argmin(sales), sales.shape)
months[row], products[column], sales[row, column]
```

```python
## Q-5: Calculate Total Sales for Each Month
```

```python
month_total_sales = np.sum(sales,axis=1)
month_total_sales
```

```python
## Q-6: Find the Highest-Selling Month
```

```python
highest_month_index = np.argmax(month_total_sales)
months[highest_month_index]
```

```python
## Q-7: Find the Lowest-Selling Month
```

```python
lowest_month_index = np.argmin(month_total_sales)
months[lowest_month_index]
```

```python
## Q-8: Calculate Total Sales for Each Product
```

```python
productwise_total_sales = np.sum(sales,axis=0)
productwise_total_sales
```

```python
## Q-9: Find the Highest-Performing Product
```

```python
highest_product_index = np.argmax(productwise_total_sales)
products[highest_product_index]
```

```python
## Q-10: Find the Lowest-Performing Product
```

```python
lowest_product_index = np.argmin(productwise_total_sales)
products[lowest_product_index]
```

```python
## Q-11: Calculate Average Sales for Each Product
```

```python
product_average_sales = np.average(sales,axis=0)
product_average_sales
```

```python
## Q-12: Calculate Average Sales for Each Month
```

```python
monthly_average_sales = np.average(sales,axis=1)
monthly_average_sales
```

```python
## Q-13: Find the Month with the Highest Average Sales
```

```python
monthly_highest_average_sales_index = np.argmax(monthly_average_sales)
months[monthly_highest_average_sales_index]
```

```python
## Q-14: Find the Month with the Lowest Average Sales
```

```python
monthly_lowest_average_sales_index = np.argmin(monthly_average_sales)
months[monthly_lowest_average_sales_index]
```

```python
## Q-15: Find the Difference Between Highest and Lowest Monthly Sales
```

```python
highest_monthly_sales = np.max(month_total_sales)
lowest_monthly_sales = np.min(month_total_sales)
difference = highest_monthly_sales - lowest_monthly_sales
difference
```

```python
## Q-16: Find the Month with the Largest Month-over-Month Sales Increase
```

```python
monthly_sales_difference = np.diff(month_total_sales)
highest_increase_index = np.argmax(monthly_sales_difference)
months[highest_increase_index + 1]
```

```python
## Q-17: Find the Product with the Highest Average Sales
```

```python
product_highest_average_sales_index = np.argmax(product_average_sales)
products[product_highest_average_sales_index]
```

```python
## Q-18: Find the Product with the Lowest Average Sales
```

```python
product_lowest_average_sales = np.argmin(product_average_sales)
products[product_lowest_average_sales]
```

```python
## Q-19: Find the Product with the Largest Sales Growth
```

```python
product_sales_growth = sales[11] - sales[0]
product_sales_growth
```

```python
highest_growth_index = np.argmax(product_sales_growth)
products[highest_growth_index]
```

```python
## Q-20: Find the Product with the Highest Sales Contribution (%)
```

```python
product_contribution = (productwise_total_sales / np.sum(sales)) * 100
highest_contribution_index = np.argmax(product_contribution)
products[highest_contribution_index]
product_contribution[highest_contribution_index]
```
